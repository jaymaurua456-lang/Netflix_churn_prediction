```python
categorical_cols = [
    'Device Used Most Often',
    'Genre Preference',
    'Region',
    'Payment History (On-Time/Delayed)',
    'Subscription Plan'
]

df = pd.get_dummies(
    df,
    columns=categorical_cols,
    drop_first=True
)
```

```python
df.head()
```

```python
df = df.drop('Customer ID', axis=1)
```

```python
df.columns
```

```python
X = df.drop('Churn Status (Yes/No)', axis=1)
```

```python
Y=df['Churn Status (Yes/No)']
```

```python
print(X.shape)
print(Y.shape)
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, random_state=42)
```

```python
print(X_train.shape)
print(X_test.shape)

print(Y_train.shape)
print(Y_test.shape)
```

```python
from sklearn.tree import DecisionTreeClassifier

dt = DecisionTreeClassifier(
    max_depth=5,
    random_state=42
)

dt.fit(X_train, Y_train)
```

```python
print(dt)
```

```python
Y_pred_dt = dt.predict(X_test)
```

```python
print(Y_pred_dt[:10])
```

```python
from sklearn.metrics import accuracy_score
accuracy_score(Y_test, Y_pred_dt)
```

```python
print(df['Churn Status (Yes/No)'].value_counts())
```

```python
from sklearn.metrics import confusion_matrix
print(confusion_matrix(Y_test, Y_pred_dt))
```

```python
from sklearn.metrics import classification_report
print(classification_report(Y_test, Y_pred_dt))
```

```python
print(dt.score(X_train, Y_train))
print(dt.score(X_test, Y_test))
```

```python
from sklearn.ensemble import RandomForestClassifier

rf = RandomForestClassifier(
    n_estimators=200,
    random_state=42
)

rf.fit(X_train, Y_train)

Y_pred_rf = rf.predict(X_test)

from sklearn.metrics import accuracy_score

print("RF Accuracy:", accuracy_score(Y_test, Y_pred_rf))
```

```python
importance_dt = pd.DataFrame({
    'Feature': X.columns,
    'Importance': dt.feature_importances_
})

importance_dt.sort_values(
    by='Importance',
    ascending=False
).head(10)
```

```python
importance_rf = pd.DataFrame({
    'Feature': X.columns,
    'Importance': rf.feature_importances_
})

importance_rf.sort_values(
    by='Importance',
    ascending=False
).head(10)
```
